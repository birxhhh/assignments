```python
import pandas as pd

# ==========================================
# PART 1: INITIAL LOADING & INSPECTION
# ==========================================
print("==========================================")
print("1. INITIAL DATA INSPECTION")
print("==========================================")

df_raw = pd.read_csv('teaching_prices_messy.csv')
df_clean = df_raw.copy()  

print("\n--- Data Info ---")
df_raw.info()
# SHOWS: Column names, non-null counts, and data types
# ACTION: Keep

print("\n--- First 10 Rows ---")
print(df_raw.head(10))
# SHOWS: A preview of the raw values to visually catch weird formatting
# ACTION: Keep

print("\n--- Missing Values Summary ---")
print(df_raw.isnull().sum())
# SHOWS: The total count of missing (NaN) values
# ACTION: Keep 

print("\n--- Full Statistical Summary ---")
print(df_raw.describe(include='all'))
# SHOWS: Statistical distribution for  columns.
# ACTION: Keep


print("\n\n==========================================")
print("2. PROBLEM 1: INCONSISTENT DATA FORMATTING")
print("==========================================")

print(df_raw['item'].unique())
# SHOWS: Every distinct variation of item names present in the column.
# ACTION: Keep 

print("\n", df_raw['item'].value_counts(dropna=False))
# SHOWS: Frequency count of each item name
# ACTION: Keep 


print("\n\n==========================================")
print("3. PROBLEM 2: MISSING VALUES IN PRICE")
print("==========================================")


missing_prices = df_raw[df_raw['price'].isna()]
print(missing_prices)
# SHOWS: The exact rows where the price value is missing (NaN).
# ACTION: Keep


print("\n\n==========================================")
print("4. PROBLEM 3: DUPLICATE RECORDS")
print("==========================================")


print("Duplicate rows count:", df_raw.duplicated().sum())
# SHOWS: Total number of completely identical row repetitions.
# ACTION: Exclude

print("\n", df_raw[df_raw.duplicated(keep=False)])
# SHOWS: All instances of duplicated rows side-by-side to verify they are true duplicates.
# ACTION: Keep


print("\n\n==========================================")
print("5. PROBLEM 4: INCORRECT DATA TYPES")
print("==========================================")


print(df_raw.dtypes)
# SHOWS: Data types of all columns, highlighting that price is an object.
# ACTION: Keep 


# Convert price to string, remove non-numeric artifacts, and cast to float
df_clean['price'] = pd.to_numeric(
    df_clean['price'].astype(str).str.replace(r'[^\d.-]', '', regex=True), 
    errors='coerce'
)

print("\n--- Cleaned Price Data Types & Stats ---")
print(df_clean['price'].dtypes)
print(df_clean['price'].describe())
# SHOWS: Confirmation that price is now a float type
# ACTION: Keep .
```
