```python
#QUESTION 1
df["product_category"].value_counts().idxmax()
#cheks the product category, counts the values and looks for the index with the most occurences, in this case, Home Decor.
```

```python
#QUESTION 3
df["city"].value_counts().idxmax()
#Same as question 1, checks the city, how many times a city appears in a row, .value_counts() sorts the cities from highest to lowest in count.
```
